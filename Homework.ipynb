### 🏋️ In-Class Exercise
Given `email = "  Student.Name@University.EDU  "`:
1. Strip whitespace
2. Convert to lowercase
3. Check if it contains `"university"` using `.find()` (returns -1 if not found)
4. Replace `".edu"` with `".com"`
5. Split on `"@"` to separate username and domain

```python
email = " Student.Name@University.EDU "

print(email.strip())
print(email.lower())

email = email.strip().lower()
print(email.find("university"))
print(email.replace(".edu", ".com"))

email = email.replace(".edu", ".com")
print(email.split("@"))

username, domain = email.split("@")
print("Username:", username)
print("Domain:", domain)
```

### 🏠 Homework Exercise
Given `sentence = "the quick brown fox jumps over the lazy dog"`:
1. Count how many times `"the"` appears
2. Capitalize the first letter of each word using `.title()`
3. Replace all spaces with underscores
4. Split the result on `"_"` and print the list
5. Join the list back with `" - "` separator

```python
sentence = "the quick brown fox jumps over the lazy dog"

print("In this sentence 'the' appears:",sentence.count("the"), "times")
print(sentence.title())
print(sentence.replace(" ","_"))

sentence = sentence.replace(" ","_")
print(sentence.split("_"))

sentence = sentence.split("_")
print("-".join(sentence))
```

Print each result with a label.

### 🏠 Homework Exercise
Create a login system simulation:
- `stored_username = "admin"`, `stored_password = "1234"`
- `input_username = "admin"`, `input_password = "wrong"`

Use boolean logic to check:
1. Is the username correct?
2. Is the password correct?
3. Is the full login successful (both correct)?
4. Should an error be shown (either incorrect)?

Print all four results.

```python
stored_username = "admin"
stored_password = "1234"

input_username = "admin"
input_password = "wrong"

print("Is the username correct?", stored_username == input_username)
print("Is the password correct?", stored_password == input_password)
print("Is the full login successful?", stored_username == input_username and stored_password == input_password)
print("Should an error be shown?", stored_username != input_username or stored_password != input_password)
```
